```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the clean  dataset and statics summary 

data = pd.read_csv("data/clean-Obesity_data.csv")
data.describe()
```

```python
#Average obesity rate by the year
# group the dataset by the year column and calculate the average of the value column and reset the index
# Then create the barplot 

yearly_avg = data.groupby("Year")["Value"].mean().reset_index()

sns.barplot(data=yearly_avg, x="Year", y="Value")
```

```python
# Obesity trends between male and female
# Filter the male and female
# Group by year and sex then calculate the average and reset the index
# create lineplot 
gender_data = data[data["Sex"].isin(["Male", "Female"])]
gender_trends = gender_data.groupby(["Year", "Sex"])["Value"].mean().reset_index()

sns.lineplot(data=gender_trends, x="Year", y="Value", hue="Sex")
```
